## 🤖 Раздел 2: Ваш первый ИИ-агент с ADK

### 🤔 2.1 Что такое ИИ-агент?

Вы, вероятно, уже использовали LLM, например, Gemini, где вы даете ему промпт, и он дает вам текстовый ответ.

`Промпт -> LLM -> Текст`

ИИ-агент делает еще один шаг вперед. Агент может думать, предпринимать действия и наблюдать за результатами этих действий, чтобы дать вам лучший ответ.

`Промпт -> Агент -> Мысль -> Действие -> Наблюдение -> Окончательный ответ`

В этом ноутбуке мы создадим агента, который может выполнять действие поиска в Google. Давайте посмотрим на разницу!

### 2.2 Определите своего агента

Теперь давайте создадим нашего агента. Мы настроим `Agent`, установив его ключевые свойства, которые сообщают ему, что делать и как работать.

Чтобы узнать больше, ознакомьтесь с документацией, связанной с [агентами в ADK](https://google.github.io/adk-docs/agents/).

Вот основные свойства, которые мы установим:

- **name** и **description**: простое имя и описание для идентификации нашего агента.
- **model**: конкретный LLM, который будет обеспечивать рассуждения агента. Мы будем использовать "gemini-2.5-flash-lite".
- **instruction**: руководящий промпт агента. Он сообщает агенту его цель и как себя вести.
- **tools**: список [инструментов](https://google.github.io/adk-docs/tools/), которые может использовать агент. Для начала мы дадим ему инструмент `google_search`, который позволяет ему находить актуальную информацию в Интернете.

````python
root_agent = Agent(
    name="helpful_assistant",
    model="gemini-1.5-flash",
    description="Простой агент, который может отвечать на общие вопросы.",
    instruction="Вы — полезный помощник. Используйте поиск Google для получения актуальной информации или если не уверены.",
    tools=[google_search],
)

print("✅ Корневой агент определен.")
````

### 2.3 Запустите своего агента

Теперь пришло время оживить вашего агента и отправить ему запрос. Для этого вам нужен [`Runner`](https://google.github.io/adk-docs/runtime/), который является центральным компонентом в ADK, действующим как оркестратор. Он управляет беседой, отправляет наши сообщения агенту и обрабатывает его ответы.

**a. Создайте `InMemoryRunner` и укажите ему использовать наш `root_agent`:**

````python
runner = InMemoryRunner(agent=root_agent)

print("✅ Runner создан.")
````

👉 Обратите внимание, что мы используем Python Runner непосредственно в этом ноутбуке. Вы также можете запускать агентов с помощью инструментов командной строки ADK, таких как `adk run`, `adk web` или `adk api_server`. Чтобы узнать больше, ознакомьтесь с документацией, связанной с [средой выполнения в ADK](https://google.github.io/adk-docs/runtime/).

**б. Теперь вы можете вызвать метод `.run_debug()`, чтобы отправить наш промпт и получить ответ.**

👉 Этот метод абстрагирует процесс создания и поддержания сеанса и используется при прототипировании. Мы рассмотрим, «что такое сеансы и как их создавать», на третий день.

````python
response = await runner.run_debug("Что такое Agent Development Kit от Google? На каких языках доступен SDK?")
````

Вы можете увидеть краткое изложение ADK и доступных языков в ответе.

### 2.4 Как это работает?

Агент выполнил поиск в Google, чтобы получить последнюю информацию о ADK, и он знал, что нужно использовать этот инструмент, потому что:

1. Агент проверяет и знает, какие инструменты ему доступны для использования.
2. В инструкциях агента указано использование инструмента поиска для получения актуальной информации или если он не уверен в ответе.

Лучший способ увидеть полную, подробную трассировку мыслей и действий агента — это **веб-интерфейс ADK**, который мы настроим позже в этом ноутбуке.

Получите свой собственный URL-адрес для доступа к веб-интерфейсу ADK в среде Kaggle Notebooks:

````python
def get_adk_proxy_url():
    """
    Получает прокси-URL для доступа к ADK из Google Colab
    """
    try:
        # Импортируем необходимый модуль
        from google.colab import output

        # Получаем URL для порта 8000 (стандартный порт ADK)
        url = output.eval_js("google.colab.kernel.proxyPort(8080)")
        return url
    except Exception as e:
        # Если не в Colab, используем localhost
        print(f"⚠️ Не удалось получить Colab proxy URL: {e}")
        return "http://localhost:8000"


url_prefix = get_adk_proxy_url()
print(f"🔗 Получен URL: {url_prefix}")
````

Теперь мы можем запустить ADK web:

````python
# ЗАПУСК GOOGLE ADK - РАБОЧИЙ ВАРИАНТ

import os
import sys
import subprocess
import threading
import time

# 1. Переходим в папку
os.chdir('/content/sample-agent')
print(f"📁 Текущая папка: {os.getcwd()}")

# 2. Проверяем файлы
print("📄 Проверяем файлы агента...")
!ls -la

# 3. Проверяем установлен ли google-adk
print("🔧 Проверяем установку google-adk...")
!python -c "import google.adk; print('✅ google-adk установлен')" 2>/dev/null || !pip install google-adk -q

# 4. Запускаем Google ADK сервер
print("\n🚀 ЗАПУСКАЕМ GOOGLE ADK СЕРВЕР...")

def run_google_adk():
    try:
        # Это правильная команда для Google ADK
        process = subprocess.Popen(
            [
                sys.executable,
                '-m', 'google.adk.runtime.webserver',
                '--host', '0.0.0.0',
                '--port', '8080'
            ],
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
            bufsize=1
        )

        # Читаем вывод в реальном времени
        for line in iter(process.stdout.readline, ''):
            print(f"[ADK] {line}", end='')

    except Exception as e:
        print(f"❌ Ошибка запуска: {e}")

# Запускаем в отдельном потоке
adk_thread = threading.Thread(target=run_google_adk, daemon=True)
adk_thread.start()

# 5. Ждем запуска
print("⏳ Ждем запуска сервера (15 секунд)...")
time.sleep(15)

# 6. Проверяем
print("\n🔍 ПРОВЕРКА ЗАПУСКА:")
print("-" * 40)

import requests
try:
    response = requests.get("http://localhost:8080", timeout=5)
    print(f"✅ HTTP статус: {response.status_code}")

    if response.status_code == 200:
        print("🎉 ADK СЕРВЕР УСПЕШНО ЗАПУЩЕН!")

        # Проверяем контент
        if len(response.text) > 100:
            print(f"📄 Получено {len(response.text)} символов HTML")

            # Ищем ключевые слова в ответе
            html_lower = response.text.lower()
            if any(keyword in html_lower for keyword in ['adk', 'agent', 'chat', 'message']):
                print("✅ Похоже на ADK интерфейс")
            else:
                print("⚠️  Контент не похож на ADK")
    else:
        print(f"⚠️  Неожиданный статус: {response.status_code}")

except requests.exceptions.ConnectionError:
    print("❌ Сервер не отвечает на localhost:8080")
    print("\n💡 Возможные причины:")
    print("   1. Сервер не запустился")
    print("   2. Запущен на другом порту")
    print("   3. Блокировка брандмауэром")
except Exception as e:
    print(f"❌ Ошибка проверки: {e}")
````
