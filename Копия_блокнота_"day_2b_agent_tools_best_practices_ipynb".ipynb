## 🧰 Раздел 2: Model Context Protocol

До сих пор вы учились создавать собственные функции для своих агентов. Но подключение к внешним системам (GitHub, базы данных, Slack) требует написания и поддержки API-клиентов.

**Model Context Protocol (MCP)** — это открытый стандарт, который позволяет агентам использовать интеграции, созданные сообществом. Вместо того чтобы писать собственные интеграции и API-клиенты, просто подключитесь к существующему MCP-серверу.

MCP позволяет агентам:

✅ **Получать доступ к актуальным внешним данным** из баз данных, API и сервисов без написания специального кода интеграции  
✅ **Использовать инструменты, созданные сообществом**, со стандартизированными интерфейсами  
✅ **Масштабировать возможности** путем подключения к нескольким специализированным серверам

### 2.1: Как работает MCP

MCP соединяет вашего агента (**клиента**) с внешними **MCP-серверами**, которые предоставляют инструменты:

- **MCP-сервер**: Предоставляет определенные инструменты (например, генерация изображений, доступ к базе данных)
- **MCP-клиент**: Ваш агент, который использует эти инструменты
- **Все серверы работают одинаково** — стандартизированный интерфейс

**Архитектура:**
```
┌──────────────────┐
│    Ваш агент     │
│   (MCP-клиент)   │
└────────┬─────────┘
         │
         │ Стандартный протокол MCP
         │
    ┌────┴────┬────────┬────────┐
    │         │        │        │
    ▼         ▼        ▼        ▼
┌────────┐ ┌─────┐ ┌──────┐ ┌─────┐
│ GitHub │ │Slack│ │ Карты│ │ ... │
│ Сервер │ │ MCP │ │  MCP │ │     │
└────────┘ └─────┘ └──────┘ └─────┘
```

### 2.2: Использование MCP с вашим агентом

Рабочий процесс прост:

1. Выберите MCP-сервер и инструмент
2. Создайте набор инструментов MCP (настройте соединение)
3. Добавьте его в свой агент
4. Запустите и протестируйте агент

**Шаг 1: Выберите MCP-сервер**

Для этой демонстрации мы будем использовать **[Everything MCP Server](https://github.com/modelcontextprotocol/servers/tree/main/src/everything)** — npm-пакет (`@modelcontextprotocol/server-everything`), предназначенный для тестирования интеграций MCP.

Он предоставляет инструмент `getTinyImage`, который возвращает простое тестовое изображение (16x16 пикселей, в кодировке Base64). **Найдите больше серверов:** [modelcontextprotocol.io/examples](https://modelcontextprotocol.io/examples)

**‼️ ПРИМЕЧАНИЕ: Это демонстрационный сервер для изучения MCP.** В производственной среде вы будете использовать серверы для Google Maps, Slack, Discord и т.д.

**Шаг 2: Создайте набор инструментов MCP**

`McpToolset` используется для интеграции агента ADK с MCP-сервером.

**Что делает код:**
- Использует `npx` (запускатель пакетов Node) для запуска MCP-сервера
- Подключается к `@modelcontextprotocol/server-everything`
- Фильтрует, чтобы использовать только инструмент `getTinyImage` (у сервера есть и другие, но нам нужен только этот)

````python
# Интеграция MCP с Everything Server
mcp_image_server = McpToolset(
    connection_params=StdioConnectionParams(
        server_params=StdioServerParameters(
            command="npx",  # Запуск MCP-сервера через npx
            args=[
                "-y",  # Аргумент для npx для автоматического подтверждения установки
                "@modelcontextprotocol/server-everything",
            ],
            tool_filter=["getTinyImage"],
        ),
        timeout=30,
    )
)

print("✅ Инструмент MCP создан")
````

#### **За кулисами:**
1. **Запуск сервера**: ADK выполняет `npx -y @modelcontextprotocol/server-everything`
2. **Handshake**: Устанавливается канал связи stdio
3. **Обнаружение инструментов**: Сервер сообщает ADK: «Я предоставляю функциональность getTinyImage»
4. **Интеграция**: Инструменты автоматически появляются в списке инструментов агента
5. **Выполнение**: Когда агент вызывает `getTinyImage()`, ADK перенаправляет вызов на MCP-сервер
6. **Ответ**: Результат от сервера бесшовно возвращается агенту

**Почему это важно**: Вы получаете мгновенный доступ к инструментам без написания кода интеграции!

**Шаг 3: Добавьте инструмент MCP в агент**

Давайте добавим `mcp_server` в массив инструментов агента и обновим инструкции агента для обработки запросов на создание крошечных изображений.

````python
# Создание агента для изображений с интеграцией MCP
image_agent = LlmAgent(
    model=Gemini(model="gemini-1.5-flash-latest", retry_options=retry_config),
    name="image_agent",
    instruction="Используйте инструмент MCP для создания изображений по запросам пользователя",
    tools=[mcp_image_server],
)
````

Создайте исполнителя (runner):

````python
from google.adk.runners import InMemoryRunner

runner = InMemoryRunner(agent=image_agent)
````

**Шаг 4: Протестируйте агент**

Попросите агента сгенерировать изображение. Посмотрите, как он использует инструмент MCP:

````python
response = await runner.run_debug("Предоставь образец крошечного изображения", verbose=True)
````

**Отобразите изображение:**

Сервер возвращает данные изображения в кодировке base64. Давайте декодируем и отобразим его:

````python
from IPython.display import display, Image as IPImage
import base64

for event in response:
    if event.content and event.content.parts:
        for part in event.content.parts:
            if hasattr(part, "function_response") and part.function_response:
                for item in part.function_response.response.get("content", []):
                    if item.get("type") == "image":
                        display(IPImage(data=base64.b64decode(item["data"])))
````

## 🔄 Раздел 3: Длительные операции (Human-in-the-Loop)

До сих пор все инструменты выполнялись и немедленно возвращали результат:


> ```Пользователь спрашивает → Агент вызывает инструмент → Инструмент возвращает результат → Агент отвечает```


**Но что, если ваши инструменты выполняются долго или вам нужно одобрение человека перед завершением действия?**

Пример: Агент по доставке должен запросить одобрение перед размещением крупного заказа.


> ```Пользователь спрашивает → Агент вызывает инструмент → Инструмент ПРИОСТАНАВЛИВАЕТСЯ и спрашивает человека → Человек одобряет → Инструмент завершается → Агент отвечает```


Это называется **длительной операцией (LRO)** — инструменту нужно приостановиться, дождаться внешнего ввода (одобрения человека), а затем возобновить работу.

**Когда использовать длительные операции:**

- 💰 **Финансовые транзакции**, требующие одобрения (переводы, покупки)
- 🗑️ **Массовые операции** (удалить 1000 записей — сначала подтвердите!)
- 📋 **Контрольные точки соответствия** (требуется регуляторное одобрение)
- 💸 **Высокозатратные действия** (запустить 50 серверов — вы уверены?)
- ⚠️ **Необратимые операции** (навсегда удалить аккаунт)

### 3.1: Что мы сегодня создаем

Давайте создадим **агента-координатора по доставке с одним инструментом**, который:
- Автоматически одобряет небольшие заказы (≤5 контейнеров)
- **Приостанавливается и запрашивает одобрение** для крупных заказов (>5 контейнеров)
- Завершает или отменяет заказ в зависимости от решения об одобрении

Это демонстрирует основной шаблон длительной операции: **приостановка → ожидание ввода от человека → возобновление**.

### 3.2: Инструмент для доставки с логикой одобрения

Вот полная функция.

#### Параметр `ToolContext`

Обратите внимание, что сигнатура функции включает `tool_context: ToolContext`. ADK автоматически предоставляет этот объект при запуске вашего инструмента. Он дает вам две ключевые возможности:

1. **Запросить одобрение:** Вызвать `tool_context.request_confirmation()`
2. **Проверить статус одобрения:** Прочитать `tool_context.tool_confirmation`

````python
LARGE_ORDER_THRESHOLD = 5


def place_shipping_order(
    num_containers: int, destination: str, tool_context: ToolContext
) -> dict:
    """Размещает заказ на доставку. Требует одобрения при заказе более 5 контейнеров (LARGE_ORDER_THRESHOLD).

    Args:
        num_containers: Количество контейнеров для отправки
        destination: Пункт назначения

    Returns:
        Словарь со статусом заказа
    """

    # -----------------------------------------------------------------------------------------------
    # -----------------------------------------------------------------------------------------------
    # СЦЕНАРИЙ 1: Небольшие заказы (≤5 контейнеров) одобряются автоматически
    if num_containers <= LARGE_ORDER_THRESHOLD:
        return {
            "status": "approved",
            "order_id": f"ORD-{num_containers}-AUTO",
            "num_containers": num_containers,
            "destination": destination,
            "message": f"Заказ автоматически одобрен: {num_containers} контейнеров в {destination}",
        }

    # -----------------------------------------------------------------------------------------------
    # -----------------------------------------------------------------------------------------------
    # СЦЕНАРИЙ 2: Этот инструмент вызывается впервые. Крупные заказы требуют одобрения человека - ПАУЗА здесь.
    if not tool_context.tool_confirmation:
        tool_context.request_confirmation(
            hint=f"⚠️ Крупный заказ: {num_containers} контейнеров в {destination}. Вы хотите одобрить?",
            payload={"num_containers": num_containers, "destination": destination},
        )
        return {  # Это отправляется Агенту
            "status": "pending",
            "message": f"Заказ на {num_containers} контейнеров требует одобрения",
        }

    # -----------------------------------------------------------------------------------------------
    # -----------------------------------------------------------------------------------------------
    # СЦЕНАРИЙ 3: Инструмент вызывается СНОВА и теперь возобновляется. Обработка ответа на одобрение - ВОЗОБНОВЛЕНИЕ здесь.
    if tool_context.tool_confirmation.confirmed:
        return {
            "status": "approved",
            "order_id": f"ORD-{num_containers}-HUMAN",
            "num_containers": num_containers,
            "destination": destination,
            "message": f"Заказ одобрен: {num_containers} контейнеров в {destination}",
        }
    else:
        return {
            "status": "rejected",
            "message": f"Заказ отклонен: {num_containers} контейнеров в {destination}",
        }


print("✅ Функции для длительных операций созданы!")
````

#### Как работают три сценария

Инструмент обрабатывает три сценария, проверяя `tool_context.tool_confirmation`:

**Сценарий 1: Небольшой заказ (≤5 контейнеров)**: Немедленно возвращает статус «автоматически одобрено».
- `tool_context.tool_confirmation` никогда не проверяется

**Сценарий 2: Крупный заказ - ПЕРВЫЙ ВЫЗОВ**
- Инструмент определяет, что это первый вызов: `if not tool_context.tool_confirmation:`
- Вызывает `request_confirmation()` для запроса одобрения у человека
- Немедленно возвращает `{'status': 'pending', ...}`
- **ADK автоматически создает событие `adk_request_confirmation`**
- Выполнение агента приостанавливается в ожидании решения человека

**Сценарий 3: Крупный заказ - ВОЗОБНОВЛЕННЫЙ ВЫЗОВ**
- Инструмент определяет, что он возобновляется: `if not tool_context.tool_confirmation:` теперь равно False
- Проверяет решение человека: `tool_context.tool_confirmation.confirmed`
- Если True → возвращает статус «одобрено»
- Если False → возвращает статус «отклонено»

**Ключевая идея:** Между двумя вызовами ваш код рабочего процесса (в Разделе 4) должен обнаружить событие `adk_request_confirmation` и возобновить работу с решением об одобрении.

### 3.4: Создание агента, приложения и исполнителя (Runner)

**Шаг 1: Создайте агента**

Добавьте инструмент в Агента. Инструмент сам решает, когда запрашивать одобрение, в зависимости от размера заказа.

````python
# Создание агента по доставке с инструментом, который можно приостановить
shipping_agent = LlmAgent(
    name="shipping_agent",
    model=Gemini(model="gemini-1.5-flash-latest", retry_options=retry_config),
    instruction="""Вы — ассистент-координатор по доставке.

  Когда пользователи запрашивают отправку контейнеров:
   1. Используйте инструмент place_shipping_order с указанием количества контейнеров и пункта назначения
   2. Если статус заказа 'pending', сообщите пользователю, что требуется одобрение
   3. После получения окончательного результата предоставьте четкое резюме, включающее:
      - Статус заказа (одобрен/отклонен)
      - ID заказа (если доступен)
      - Количество контейнеров и пункт назначения
   4. Ответы должны быть краткими, но информативными
  """,
    tools=[FunctionTool(func=place_shipping_order)],
)

print("✅ Агент по доставке создан!")
````

**Шаг 2: Оберните в возобновляемое приложение (App)**

**Проблема:** Обычный `LlmAgent` не имеет состояния — каждый вызов независим и не помнит предыдущих взаимодействий. Если инструмент запрашивает одобрение, агент не может вспомнить, что он делал.

**Решение:** Оберните вашего агента в **`App`** с **включенной возможностью возобновления**. Приложение добавляет слой персистентности, который сохраняет и восстанавливает состояние.

**Что сохраняется при приостановке инструмента:**
- Все сообщения диалога до этого момента
- Какой инструмент был вызван (`place_shipping_order`)
- Параметры инструмента (10 контейнеров, Роттердам)
- Точное место, где он приостановился (в ожидании одобрения)

Когда вы возобновляете работу, приложение загружает это сохраненное состояние, чтобы агент продолжил с того места, где остановился, как будто время и не проходило.

````python
# Оберните агента в возобновляемое приложение - ЭТО КЛЮЧ К ДЛИТЕЛЬНЫМ ОПЕРАЦИЯМ!
shipping_app = App(
    name="shipping_coordinator",
    root_agent=shipping_agent,
    resumability_config=ResumabilityConfig(is_resumable=True),
)

print("✅ Возобновляемое приложение создано!")
````

**Шаг 3: Создайте сессию и исполнителя (Runner) с приложением**

Передайте `app=shipping_app` вместо `agent=...`, чтобы исполнитель знал о возможности возобновления.

````python
session_service = InMemorySessionService()

# Создайте исполнителя с возобновляемым приложением
shipping_runner = Runner(
    app=shipping_app,  # Передайте приложение вместо агента
    session_service=session_service,
)

print("✅ Исполнитель создан!")
````

## 🏗️ Раздел 4: Создание рабочего процесса

‼️ **Важно:** Код рабочего процесса использует концепции ADK, такие как сессии, исполнители и события. **Мы рассмотрим то, что вам нужно знать для длительных операций** в этом блокноте. Для более глубокого понимания мы рассмотрим эти темы на 3-й день, или вы можете ознакомиться с [документацией ADK](https://google.github.io/adk-docs/runtime/) и этим [видео](https://www.youtube.com/watch?v=44C8u0CDtSo&list=PLOU2XLYxmsIIAPgM8FmtEcFTXLLzmh4DK&index=2&t=1s).

### 4.1: ⚠️ Критическая часть - Обработка событий в вашем рабочем процессе

Агент не будет автоматически обрабатывать приостановку/возобновление. **Каждый рабочий процесс с длительными операциями требует от вас:**

1. **Обнаружить приостановку:** Проверить, содержат ли события `adk_request_confirmation`
2. **Получить решение человека:** В производственной среде показать UI и дождаться клика пользователя. Здесь мы это симулируем.
3. **Возобновить агента:** Отправить решение обратно с сохраненным `invocation_id`

### 4.2 Понимание ключевых технических концепций

👉 **`events`** - ADK создает события по мере выполнения агента. Вызовы инструментов, ответы модели, результаты функций - все это становится событиями

👉 **Событие `adk_request_confirmation`** - Это событие особенное - оно сигнализирует «приостановить здесь!»
- Автоматически создается ADK, когда ваш инструмент вызывает `request_confirmation()`
- Содержит `invocation_id`
- Ваш рабочий процесс должен обнаружить это событие, чтобы знать, что агент приостановился

👉 **`invocation_id`** - Каждый вызов `run_async()` получает уникальный `invocation_id` (например, «abc123»)
- Когда инструмент приостанавливается, вы сохраняете этот ID
- При возобновлении передайте тот же ID, чтобы ADK знал, какое выполнение продолжить
- Без него ADK начал бы НОВОЕ выполнение вместо возобновления приостановленного

### 4.3: Вспомогательные функции для обработки событий

Они берут на себя логику итерации по событиям.

**`check_for_approval()`** - Обнаруживает, приостановился ли агент
- Проходит по всем событиям и ищет специальное событие `adk_request_confirmation`
- Возвращает `approval_id` (идентифицирует этот конкретный запрос) и `invocation_id` (идентифицирует, какое выполнение возобновить)
- Возвращает `None`, если приостановка не обнаружена

````python
def check_for_approval(events):
    """Проверяет, содержат ли события запрос на одобрение.

    Returns:
        Словарь с деталями одобрения или None
    """
    for event in events:
        if event.content and event.content.parts:
            for part in event.content.parts:
                if (
                    part.function_call
                    and part.function_call.name == "adk_request_confirmation"
                ):
                    return {
                        "approval_id": part.function_call.id,
                        "invocation_id": event.invocation_id,
                    }
    return None
````

**`print_agent_response()`** - Отображает текст агента
- Простая вспомогательная функция для извлечения и печати текста из событий

````python
def print_agent_response(events):
    """Печатает текстовые ответы агента из событий."""
    for event in events:
        if event.content and event.content.parts:
            for part in event.content.parts:
                if part.text:
                    print(f"Агент > {part.text}")
````

**`create_approval_response()`** - Форматирует решение человека
- Принимает информацию об одобрении и булево решение (True/False) от человека
- Создает `FunctionResponse`, который понимает ADK
- Оборачивает его в объект `Content` для отправки обратно агенту

````python
def create_approval_response(approval_info, approved):
    """Создает сообщение с ответом на одобрение."""
    confirmation_response = types.FunctionResponse(
        id=approval_info["approval_id"],
        name="adk_request_confirmation",
        response={"confirmed": approved},
    )
    return types.Content(
        role="user", parts=[types.Part(function_response=confirmation_response)]
    )


print("✅ Вспомогательные функции определены")
````

### 4.4: Функция рабочего процесса - Давайте свяжем все воедино!

Функция `run_shipping_workflow()` организует весь процесс одобрения.

Посмотрите объяснение кода в ячейке ниже.

<img src="https://storage.googleapis.com/github-repo/kaggle-5days-ai/day2/lro-workflow.png" width="1000" alt="Рабочий процесс длительной операции">

````python
async def run_shipping_workflow(query: str, auto_approve: bool = True):
    """Запускает рабочий процесс доставки с обработкой одобрения.

    Args:
        query: Запрос пользователя на доставку
        auto_approve: Автоматически ли одобрять крупные заказы (симулирует решение человека)
    """

    print(f"\n{'='*60}")
    print(f"Пользователь > {query}\n")

    # Генерация уникального ID сессии
    session_id = f"order_{uuid.uuid4().hex[:8]}"

    # Создание сессии
    await session_service.create_session(
        app_name="shipping_coordinator", user_id="test_user", session_id=session_id
    )

    query_content = types.Content(role="user", parts=[types.Part(text=query)])
    events = []

    # -----------------------------------------------------------------------------------------------
    # -----------------------------------------------------------------------------------------------
    # ШАГ 1: Отправка первоначального запроса Агенту. Если num_containers > 5, Агент возвращает специальное событие `adk_request_confirmation`
    async for event in shipping_runner.run_async(
        user_id="test_user", session_id=session_id, new_message=query_content
    ):
        events.append(event)

    # -----------------------------------------------------------------------------------------------
    # -----------------------------------------------------------------------------------------------
    # ШАГ 2: Пройдитесь по всем сгенерированным событиям и проверьте наличие `adk_request_confirmation`.
    approval_info = check_for_approval(events)

    # -----------------------------------------------------------------------------------------------
    # -----------------------------------------------------------------------------------------------
    # ШАГ 3: Если событие присутствует, это крупный заказ - ОБРАБОТКА ПРОЦЕССА ОДОБРЕНИЯ
    if approval_info:
        print(f"⏸️  Приостановка для одобрения...")
        print(f"🤔 Решение человека: {'ОДОБРИТЬ ✅' if auto_approve else 'ОТКЛОНИТЬ ❌'}\n")

        # ПУТЬ А: Возобновите работу агента, вызвав run_async() снова с решением об одобрении
        async for event in shipping_runner.run_async(
            user_id="test_user",
            session_id=session_id,
            new_message=create_approval_response(
                approval_info, auto_approve
            ),  # Отправьте решение человека здесь
            invocation_id=approval_info[
                "invocation_id"
            ],  # Критически важно: тот же invocation_id говорит ADK ВОЗОБНОВИТЬ
        ):
            if event.content and event.content.parts:
                for part in event.content.parts:
                    if part.text:
                        print(f"Агент > {part.text}")

    # -----------------------------------------------------------------------------------------------
    # -----------------------------------------------------------------------------------------------
    else:
        # ПУТЬ Б: Если `adk_request_confirmation` отсутствует - одобрение не требуется - заказ выполнен немедленно.
        print_agent_response(events)

    print(f"{'='*60}\n")


print("✅ Функция рабочего процесса готова")
````

#### **Разбор кода**

**Шаг 1: Отправка первоначального запроса Агенту**
- Вызовите `run_async()` для запуска выполнения агента
- Соберите все события в список для проверки

**Шаг 2: Обнаружение приостановки**
- Вызовите `check_for_approval(events)` для поиска специального события: `adk_request_confirmation`
- Возвращает информацию об одобрении (с `invocation_id`), если специальное событие присутствует; `None`, если выполнено

**Шаг 3: Возобновление выполнения**

ПУТЬ А:
- Если информация об одобрении присутствует, в этот момент Агент *приостанавливается* для получения ввода от человека.
- Как только ввод от человека доступен, вызовите агента снова с помощью `run_async()` и передайте ввод от человека.
- **Критически важно:** тот же `invocation_id` (говорит ADK ВОЗОБНОВИТЬ, а не перезапускать)
- Отобразите окончательный ответ агента после возобновления

ПУТЬ Б:
- Если информация об одобрении отсутствует, то одобрение не требуется, и агент завершает выполнение.

### 🎬 4.5: Демонстрация: Тестирование рабочего процесса

Теперь давайте запустим наши демонстрации. Обратите внимание, насколько они стали чище и легче для чтения. Вся сложная логика приостановки и возобновления теперь скрыта в нашей вспомогательной функции `run_workflow`, что позволяет нам сосредоточиться на задачах, которые мы хотим, чтобы агент выполнял.

**Примечание:** Вы можете видеть предупреждения типа `Warning: there are non-text parts in the response: ['function_call']` — это нормально и можно игнорировать. Это просто означает, что агент вызывает инструменты в дополнение к генерации текста.

````python
# Демо 1: Это небольшой заказ. Агент получает от инструмента статус «автоматически одобрено»
await run_shipping_workflow("Отправить 3 контейнера в Сингапур")

# Демо 2: Рабочий процесс симулирует решение человека: ОДОБРИТЬ ✅
await run_shipping_workflow("Отправить 10 контейнеров в Роттердам", auto_approve=True)

# Демо 3: Рабочий процесс симулирует решение человека: ОТКЛОНИТЬ ❌
await run_shipping_workflow("Отправить 8 контейнеров в Лос-Анджелес", auto_approve=False)
````

### 4.6: (Необязательно) Полный поток выполнения

Вот пример трассировки всего рабочего процесса.

**TL;DR:** Инструмент приостанавливается во ВРЕМЯ 6, рабочий процесс обнаруживает это во ВРЕМЯ 8, возобновляется во ВРЕМЯ 10 с тем же `invocation_id="abc123"`.

**Подробная временная шкала:**

Вот что происходит шаг за шагом, когда вы запускаете `run_shipping_workflow("Отправить 10 контейнеров в Роттердам", auto_approve=True)`:

```
ВРЕМЯ 1: Пользователь отправляет «Отправить 10 контейнеров в Роттердам»
        ↓
ВРЕМЯ 2: Рабочий процесс вызывает shipping_runner.run_async(...)
        ADK присваивает уникальный invocation_id = «abc123»
        ↓
ВРЕМЯ 3: Агент получает сообщение пользователя, решает использовать инструмент place_shipping_order
        ↓
ВРЕМЯ 4: ADK вызывает place_shipping_order(10, «Роттердам», tool_context)
        ↓
ВРЕМЯ 5: Инструмент проверяет: num_containers (10) > 5
        Инструмент вызывает tool_context.request_confirmation(...)
        ↓
ВРЕМЯ 6: Инструмент возвращает {'status': 'pending', ...}
        ↓
ВРЕМЯ 7: ADK создает событие adk_request_confirmation с invocation_id=«abc123»
        ↓
ВРЕМЯ 8: Рабочий процесс обнаруживает событие через check_for_approval()
        Сохраняет approval_id и invocation_id=«abc123»
        ↓
ВРЕМЯ 9: Рабочий процесс получает решение человека → True (одобрить)
        ↓
ВРЕМЯ 10: Рабочий процесс вызывает shipping_runner.run_async(..., invocation_id=«abc123»)
         Передает решение об одобрении как FunctionResponse
         ↓
ВРЕМЯ 11: ADK видит invocation_id=«abc123» — знает, что нужно ВОЗОБНОВИТЬ (а не начинать заново)
         Загружает сохраненное состояние из ВРЕМЕНИ 7
         ↓
ВРЕМЯ 12: ADK вызывает place_shipping_order снова с теми же параметрами
         Но теперь tool_context.tool_confirmation.confirmed = True
         ↓
ВРЕМЯ 13: Инструмент возвращает {'status': 'approved', 'order_id': 'ORD-10-HUMAN', ...}
         ↓
ВРЕМЯ 14: Агент получает результат и отвечает пользователю
```

**Ключевой момент:** `invocation_id` — это то, как ADK узнает, что нужно возобновить приостановленное выполнение, а не начинать новое.

| **Интеграция MCP** | Вам нужно подключиться к **внешним, стандартизированным сервисам** (таким как время, базы данных или файловые системы) без написания специального кода интеграции. | `McpToolset` |
| **Длительные операции** | Вам нужно **приостановить рабочий процесс**, чтобы дождаться внешнего события, чаще всего для **одобрения человеком (human-in-the-loop)**, длительных фоновых задач или для контрольных точек соответствия/безопасности. | `ToolContext`, `request_confirmation`, `App`, `ResumabilityConfig` |

- ⏳ **Управляют временем**: Управляют операциями, которые могут длиться минуты, часы или дни  
- 🔒 **Обеспечивают соответствие**: Добавляют человеческий надзор к критически важным операциям
- 🔄 **Сохраняют состояние**: Возобновляют диалоги именно там, где они были приостановлены
