### 1.4: Вспомогательные функции

Вспомогательная функция, которая выводит сгенерированный код Python и результаты выполнения кода из инструмента выполнения кода:

````python
def show_python_code_and_result(response):
    for i in range(len(response)):
        # Check if the response contains a valid function call result from the code executor
        if (
            (response[i].content.parts)
            and (response[i].content.parts[0])
            and (response[i].content.parts[0].function_response)
            and (response[i].content.parts[0].function_response.response)
        ):
            response_code = response[i].content.parts[0].function_response.response
            if "result" in response_code and response_code["result"] != "```":
                if "tool_code" in response_code["result"]:
                    print(
                        "Generated Python Code >> ",
                        response_code["result"].replace("tool_code", ""),
                    )
                else:
                    print("Generated Python Response >> ", response_code["result"])


print("✅ Helper functions defined.")
````

## 🤖 Раздел 2: Что такое пользовательские инструменты?

**Пользовательские инструменты** — это инструменты, которые вы создаете сами, используя собственный код и бизнес-логику. В отличие от встроенных инструментов, которые поставляются готовыми с ADK, пользовательские инструменты дают вам полный контроль над функциональностью.

**Когда использовать пользовательские инструменты?**

Встроенные инструменты, такие как Google Search, мощные, но **у каждого бизнеса есть уникальные требования**, которые не могут быть выполнены с помощью общих инструментов. Пользовательские инструменты позволяют вам реализовать вашу специфическую бизнес-логику, подключаться к вашим системам и решать проблемы, специфичные для вашей предметной области. ADK предоставляет несколько типов пользовательских инструментов для обработки этих сценариев.

### 2.1: Создание пользовательских функциональных инструментов

#### Пример: Агент для конвертации валют

Этот агент может конвертировать валюту из одной деноминации в другую и рассчитывать комиссию за конвертацию. Агент имеет два пользовательских инструмента и следует рабочему процессу:

1. **Инструмент для поиска комиссии** - находит комиссию за транзакцию для конвертации (имитация)
2. **Инструмент для получения курса обмена** - получает курсы обмена валют (имитация)
3. **Шаг расчета** - рассчитывает общую стоимость конвертации, включая комиссии

### 🤔 2.2: Как определить Инструмент?

**Любая функция Python может стать инструментом агента**, следуя этим простым рекомендациям:

1. Создайте функцию Python
2. Следуйте лучшим практикам, перечисленным ниже
3. Добавьте вашу функцию в список `tools=[]` агента, и ADK автоматически сделает все остальное.


#### 🏆 Лучшие практики ADK в действии

Обратите внимание, как наши инструменты следуют лучшим практикам ADK:

**1. Возвращаемые значения в виде словаря**: Инструменты возвращают `{"status": "success", "data": ...}` или `{"status": "error", "error_message": ...}`  
**2. Четкие докстринги**: LLM используют докстринги, чтобы понять, когда и как использовать инструменты  
**3. Указания типов**: Позволяют ADK генерировать правильные схемы (`str`, `dict` и т. д.)  
**4. Обработка ошибок**: Структурированные ответы об ошибках помогают LLM корректно обрабатывать сбои  

Эти шаблоны делают ваши инструменты надежными и простыми для правильного использования LLM.

👉 Давайте посмотрим на это в действии на примере нашего первого инструмента:

````python
# Pay attention to the docstring, type hints, and return value.
def get_fee_for_payment_method(method: str) -> dict:
    """Looks up the transaction fee percentage for a given payment method.

    This tool simulates looking up a company's internal fee structure based on
    the name of the payment method provided by the user.

    Args:
        method: The name of the payment method. It should be descriptive,
                e.g., "platinum credit card" or "bank transfer".

    Returns:
        Dictionary with status and fee information.
        Success: {"status": "success", "fee_percentage": 0.02}
        Error: {"status": "error", "error_message": "Payment method not found"}
    """
    # This simulates looking up a company's internal fee structure.
    fee_database = {
        "platinum credit card": 0.02,  # 2%
        "gold debit card": 0.035,  # 3.5%
        "bank transfer": 0.01,  # 1%
    }

    fee = fee_database.get(method.lower())
    if fee is not None:
        return {"status": "success", "fee_percentage": fee}
    else:
        return {
            "status": "error",
            "error_message": f"Payment method '{method}' not found",
        }


print("✅ Fee lookup function created")
print(f"💳 Test: {get_fee_for_payment_method('platinum credit card')}")
````

Давайте будем следовать тем же лучшим практикам для определения нашего второго инструмента `get_exchange_rate`.

````python
def get_exchange_rate(base_currency: str, target_currency: str) -> dict:
    """Looks up and returns the exchange rate between two currencies.

    Args:
        base_currency: The ISO 4217 currency code of the currency you
                       are converting from (e.g., "USD").
        target_currency: The ISO 4217 currency code of the currency you
                         are converting to (e.g., "EUR").

    Returns:
        Dictionary with status and rate information.
        Success: {"status": "success", "rate": 0.93}
        Error: {"status": "error", "error_message": "Unsupported currency pair"}
    """

    # Static data simulating a live exchange rate API
    # In production, this would call something like: requests.get("api.exchangerates.com")
    rate_database = {
        "usd": {
            "eur": 0.93,  # Euro
            "jpy": 157.50,  # Japanese Yen
            "inr": 83.58,  # Indian Rupee
        }
    }

    # Input validation and processing
    base = base_currency.lower()
    target = target_currency.lower()

    # Return structured result with status
    rate = rate_database.get(base, {}).get(target)
    if rate is not None:
        return {"status": "success", "rate": rate}
    else:
        return {
            "status": "error",
            "error_message": f"Unsupported currency pair: {base_currency}/{target_currency}",
        }


print("✅ Exchange rate function created")
print(f"💱 Test: {get_exchange_rate('USD', 'EUR')}")
````

 Теперь давайте создадим нашего агента по обмену валют. Обратите внимание, как инструкции агента ссылаются на инструменты:

**Ключевые моменты:**
- Список `tools=[]` сообщает агенту, какие функции он может использовать
- Инструкции ссылаются на инструменты по их точным именам функций (например,
`get_fee_for_payment_method()`)
- Агент использует эти имена, чтобы решить, когда и как вызывать каждый инструмент

````python
# Currency agent with custom function tools
currency_agent = LlmAgent(
    name="currency_agent",
    model=Gemini(model="gemini-2.5-flash-lite", retry_options=retry_config),
    instruction="""You are a smart currency conversion assistant.

    For currency conversion requests:
    1. Use `get_fee_for_payment_method()` to find transaction fees
    2. Use `get_exchange_rate()` to get currency conversion rates
    3. Check the "status" field in each tool's response for errors
    4. Calculate the final amount after fees based on the output from `get_fee_for_payment_method` and `get_exchange_rate` methods and provide a clear breakdown.
    5. First, state the final converted amount.
        Then, explain how you got that result by showing the intermediate amounts. Your explanation must include: the fee percentage and its
        value in the original currency, the amount remaining after the fee, and the exchange rate used for the final conversion.

    If any tool returns status "error", explain the issue to the user clearly.
    """,
    tools=[get_fee_for_payment_method, get_exchange_rate],
)

print("✅ Currency agent created with custom function tools")
print("🔧 Available tools:")
print("  • get_fee_for_payment_method - Looks up company fee structure")
print("  • get_exchange_rate - Gets current exchange rates")
````

````python
# Test the currency agent
currency_runner = InMemoryRunner(agent=currency_agent)
_ = await currency_runner.run_debug(
    "I want to convert 500 US Dollars to Euros using my Platinum Credit Card. How much will I receive?"
)
````

**Отлично!** Наш агент теперь использует пользовательскую бизнес-логику со структурированными ответами.

## 💻 Раздел 3: Повышение надежности агента с помощью кода

В инструкции агента говорится *"рассчитать конечную сумму после вычета комиссий"*, но LLM не всегда надежны в математике. Они могут делать ошибки в расчетах или использовать несогласованные формулы.

##### 💡 **Решение:** Давайте попросим нашего агента сгенерировать код на Python для выполнения математических расчетов и запустить его, чтобы получить окончательный результат! Выполнение кода гораздо надежнее, чем попытки LLM выполнить математические расчеты в уме!

<img src="https://storage.googleapis.com/github-repo/kaggle-5days-ai/day2/enhanced-currency-agent.png" width="800" alt="Усовершенствованный агент-конвертер валют">

### 3.1 Встроенный исполнитель кода

В ADK есть встроенный исполнитель кода, способный запускать код в песочнице. **Примечание:** Это использует возможности Gemini по выполнению кода.

Давайте создадим `calculation_agent`, который принимает код Python и использует `BuiltInCodeExecutor` для его запуска.

````python
calculation_agent = LlmAgent(
    name="CalculationAgent",
    model=Gemini(model="gemini-2.5-flash-lite", retry_options=retry_config),
    instruction="""You are a specialized calculator that ONLY responds with Python code. You are forbidden from providing any text, explanations, or conversational responses.

     Your task is to take a request for a calculation and translate it into a single block of Python code that calculates the answer.

     **RULES:**
    1.  Your output MUST be ONLY a Python code block.
    2.  Do NOT write any text before or after the code block.
    3.  The Python code MUST calculate the result.
    4.  The Python code MUST print the final result to stdout.
    5.  You are PROHIBITED from performing the calculation yourself. Your only job is to generate the code that will perform the calculation.

    Failure to follow these rules will result in an error.
       """,
    code_executor=BuiltInCodeExecutor(),  # Use the built-in Code Executor Tool. This gives the agent code execution capabilities
)
````

### 3.2: Обновление инструкций и набора инструментов Агента

Мы выполним два ключевых действия:

1. **Обновим инструкции `currency_agent` для генерации кода на Python**
- Исходные: "*Рассчитать итоговую сумму после уплаты комиссий*" (нечеткие математические инструкции)
- Улучшенные: "*Сгенерировать код на Python для расчета итоговой суммы... и использовать `calculation_agent` для выполнения кода и вычисления итоговой суммы*"

2. **Добавим `calculation_agent` в набор инструментов**

    ADK позволяет использовать любого агента в качестве инструмента с помощью `AgentTool`.

- Добавьте `AgentTool(agent=calculation_agent)` в список инструментов
- Специализированный агент становится доступным для вызова корневым агентом

Давайте посмотрим, как это работает на практике:

````python
enhanced_currency_agent = LlmAgent(
    name="enhanced_currency_agent",
    model=Gemini(model="gemini-2.5-flash-lite", retry_options=retry_config),
    # Updated instruction
    instruction="""You are a smart currency conversion assistant. You must strictly follow these steps and use the available tools.

  For any currency conversion request:

   1. Get Transaction Fee: Use the get_fee_for_payment_method() tool to determine the transaction fee.
   2. Get Exchange Rate: Use the get_exchange_rate() tool to get the currency conversion rate.
   3. Error Check: After each tool call, you must check the "status" field in the response. If the status is "error", you must stop and clearly explain the issue to the user.
   4. Calculate Final Amount (CRITICAL): You are strictly prohibited from performing any arithmetic calculations yourself. You must use the calculation_agent tool to generate Python code that calculates the final converted amount. This
      code will use the fee information from step 1 and the exchange rate from step 2.
   5. Provide Detailed Breakdown: In your summary, you must:
       * State the final converted amount.
       * Explain how the result was calculated, including:
           * The fee percentage and the fee amount in the original currency.
           * The amount remaining after deducting the fee.
           * The exchange rate applied.
    """,
    tools=[
        get_fee_for_payment_method,
        get_exchange_rate,
        AgentTool(agent=calculation_agent),  # Using another agent as a tool!
    ],
)

print("✅ Enhanced currency agent created")
print("🎯 New capability: Delegates calculations to specialist agent")
print("🔧 Tool types used:")
print("  • Function Tools (fees, rates)")
print("  • Agent Tool (calculation specialist)")
````

````python
# Define a runner
enhanced_runner = InMemoryRunner(agent=enhanced_currency_agent)
````

````python
# Test the enhanced agent
response = await enhanced_runner.run_debug(
    "Convert 1,250 USD to INR using a Bank Transfer. Show me the precise calculation."
)
````

**Отлично!** Обратите внимание, что произошло:

- Когда агент по обмену валют вызывает `CalculationAgent`, он передает ему сгенерированный код Python.

### 🤔 3.3: Инструменты Агента и Субагенты: В чем разница?

Это частый вопрос! Оба подхода включают использование нескольких агентов, но они работают совершенно по-разному:

**Инструменты Агента (то, что мы используем):**
- Агент А вызывает Агента Б как инструмент
- Ответ Агента Б возвращается **обратно к Агенту А**
- Агент А сохраняет контроль и продолжает диалог
- **Сценарий использования**: Делегирование для конкретных задач (например, расчетов)

**Субагенты (другой шаблон):**
- Агент А передает контроль **полностью Агенту Б**
- Агент Б берет на себя управление и обрабатывает все последующие вводы пользователя
- Агент А выходит из процесса
- **Сценарий использования**: Передача задач специалистам (например, по уровням поддержки клиентов)

**В нашем примере с валютой:** Мы хотим, чтобы агент по обмену валют получал результаты расчетов и продолжал работать с ними, поэтому мы используем **Инструменты Агента**, а не субагентов.

## 🧰 Раздел 4: Полное руководство по типам инструментов ADK

Теперь, когда вы увидели инструменты в действии, давайте разберемся с полным набором инструментов ADK:

Он условно делится на две категории: **Пользовательские инструменты** и **Встроенные инструменты**

### **1. Пользовательские инструменты**

<img src="https://storage.googleapis.com/github-repo/kaggle-5days-ai/day2/custom-tools.png" width="800" alt="Пользовательские инструменты">

**Что это**: Инструменты, которые вы создаете сами для конкретных нужд

**Преимущество**: Полный контроль над функциональностью — вы создаете именно то, что нужно вашему агенту

#### **Функциональные инструменты** ✅ (Вы их уже использовали!)
- **Что это**: Функции Python, преобразованные в инструменты агента
- **Примеры**: `get_fee_for_payment_method`, `get_exchange_rate`
- **Преимущество**: Мгновенно превращайте любую функцию Python в инструмент агента

#### **Долго выполняющиеся функциональные инструменты**
- **Что это**: Функции для операций, которые занимают значительное время
- **Примеры**: Утверждения с участием человека, обработка файлов
- **Преимущество**: Агенты могут запускать задачи и продолжать другую работу в ожидании

#### **Инструменты-агенты** ✅ (Вы их уже использовали!)
- **Что это**: Другие агенты, используемые в качестве инструментов
- **Примеры**: `AgentTool(agent=calculation_agent)`
- **Преимущество**: Создавайте специализированных агентов и повторно используйте их в разных системах

#### **Инструменты MCP**
- **Что это**: Инструменты с серверов Model Context Protocol
- **Примеры**: Доступ к файловой системе, Google Maps, базам данных
- **Преимущество**: Подключайтесь к любому MCP-совместимому сервису без специальной интеграции

#### **Инструменты OpenAPI**
- **Что это**: Инструменты, автоматически генерируемые из спецификаций API
- **Примеры**: Конечные точки REST API становятся вызываемыми инструментами
- **Преимущество**: Никакого ручного кодирования — просто предоставьте спецификацию API и получите рабочие инструменты

### **2. Встроенные инструменты**

<img src="https://storage.googleapis.com/github-repo/kaggle-5days-ai/day2/built-in-tools.png" width="1200" alt="Встроенные инструменты">

**Что это**: Готовые инструменты, предоставляемые ADK

**Преимущество**: Нет времени на разработку — используйте немедленно без настройки

#### **Инструменты Gemini** ✅ (Вы их уже использовали!)
- **Что это**: Инструменты, использующие возможности Gemini
- **Примеры**: `google_search`, `BuiltInCodeExecutor`
- **Преимущество**: Надежные, протестированные инструменты, которые работают «из коробки»

#### **Инструменты Google Cloud** [требуется доступ к Google Cloud]
- **Что это**: Инструменты для сервисов Google Cloud и корпоративной интеграции
- **Примеры**: `BigQueryToolset`, `SpannerToolset`, `APIHubToolset`
- **Преимущество**: Доступ к базам данных и API корпоративного уровня со встроенной безопасностью

#### **Сторонние инструменты**
- **Что это**: Оболочки для существующих экосистем инструментов
- **Примеры**: Hugging Face, Firecrawl, GitHub Tools
- **Преимущество**: Повторное использование существующих инвестиций в инструменты — не нужно пересоздавать то, что уже существует
