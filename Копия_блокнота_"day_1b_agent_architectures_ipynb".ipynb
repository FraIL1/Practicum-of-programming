**Раздел 2: Введение в многоагентные системы**

>
>
> -----
>
> ### Раздел 2
>
> ## 🤔 Зачем нужны многоагентные системы? + Ваш первый мульти-агент
>
> **Проблема: Агент «Мастер на все руки»**
>
> Одиночные агенты могут многое. Но что происходит, когда задача становится сложной? Единый «монолитный» агент, который пытается одновременно заниматься исследованиями, написанием текста, редактированием и проверкой фактов, становится проблемой. Его инструкция (промпт) становится длинной и запутанной. Его трудно отлаживать (какая часть дала сбой?), сложно поддерживать, и он часто выдает ненадежные результаты.
>
> **Решение: Команда специалистов**
>
> Вместо одного агента «мастера на все руки» мы можем построить **многоагентную систему**. Это команда простых, специализированных агентов, которые сотрудничают, как и команда в реальном мире. У каждого агента есть одна четкая задача (например, один агент *только* проводит исследования, другой *только* пишет). Это делает их проще в создании, проще в тестировании, а также намного мощнее и надежнее при совместной работе.
>
> Чтобы узнать больше, ознакомьтесь с документацией по [LLM-агентам в ADK](https://google.github.io/adk-docs/agents/llm-agents/).
>
> **Архитектура: Одиночный агент против Многоагентной команды**

-----

#### Создание Агента-Исследователя и Агента-Резюматора

````
# Агент-исследователь: Его работа — использовать инструмент google_search и представлять результаты.
research_agent = Agent(
    name="ResearchAgent", # Имя агента
    model="gemini-2.5-flash-lite",
    instruction="""Вы — специализированный агент-исследователь. Ваша единственная работа — использовать
    инструмент google_search, чтобы найти 2-3 части релевантной информации по заданной теме и представить результаты с указанием источников.""",
    tools=[google_search],
    output_key="research_findings", # Результат работы этого агента будет сохранен в состоянии сессии с этим ключом.
)

print("✅ research_agent создан.")
````

````
# Агент-резюматор: Его работа — обобщать текст, который он получает.
summarizer_agent = Agent(
    name="SummarizerAgent",
    model="gemini-2.5-flash-lite",
    # Инструкция изменена для запроса маркированного списка для более четкого формата вывода.
    instruction="""Прочитайте предоставленные результаты исследования: {research_findings}
Создайте краткое резюме в виде маркированного списка из 3-5 ключевых пунктов.""",
    output_key="final_summary",
)

print("✅ summarizer_agent создан.")
````

-----

**Координация агентов**

> Ознакомьтесь с документацией ADK для получения дополнительной информации о том, как [управлять агентами с помощью четких и конкретных инструкций](https://google.github.io/adk-docs/agents/llm-agents/).
>
> Затем мы объединяем агентов под управлением корневого агента или координатора:

````
# Главный координатор: Организует рабочий процесс, вызывая субагентов как инструменты.
root_agent = Agent(
    name="ResearchCoordinator",
    model="gemini-2.5-flash-lite",
    # Эта инструкция говорит главному агенту, КАК использовать свои инструменты (которыми являются другие агенты).
    instruction="""Вы — координатор исследований. Ваша цель — ответить на запрос пользователя, организовав рабочий процесс.
1. Сначала вы ДОЛЖНЫ вызвать инструмент `ResearchAgent`, чтобы найти релевантную информацию по теме, предоставленной пользователем.
2. Затем, после получения результатов исследования, вы ДОЛЖНЫ вызвать инструмент `SummarizerAgent`, чтобы создать краткое резюме.
3. Наконец, четко представьте итоговое резюме пользователю в качестве вашего ответа.""",
    # Мы оборачиваем субагентов в `AgentTool`, чтобы сделать их вызываемыми инструментами для главного агента.
    tools=[
        AgentTool(research_agent),
        AgentTool(summarizer_agent)
    ],
)

print("✅ root_agent создан.")
````

> Здесь мы используем `AgentTool`, чтобы обернуть субагентов и сделать их вызываемыми инструментами для главного агента. Мы подробно рассмотрим `AgentTool` во 2-й день.
>
> Давайте запустим агента и спросим его о какой-нибудь теме:

````
runner = InMemoryRunner(agent=root_agent)
response = await runner.run_debug("Каковы последние достижения в квантовых вычислениях и что они значат для ИИ?")
````

> Вы только что построили свою первую многоагентную систему\! Вы использовали одного агента-«координатора» для управления рабочим процессом, что является мощным и гибким шаблоном.
>
> ‼️ Однако **полагаться на инструкции LLM для контроля порядка действий иногда может быть непредсказуемым.** Далее мы рассмотрим другой шаблон, который дает вам гарантированное пошаговое выполнение.

**Раздел 3: Последовательные рабочие процессы**

>
>
> -----
>
> ### Раздел 3
>
> ## 🚥 Последовательные рабочие процессы (Sequential) — Конвейер
>
> **Проблема: Непредсказуемый порядок**
>
> Предыдущая многоагентная система работала, но она полагалась на **подробный промпт с инструкциями**, чтобы заставить LLM выполнять шаги по порядку. Это может быть ненадежно. Сложная LLM может решить пропустить шаг, выполнить их в неправильном порядке или «застрять», делая процесс непредсказуемым.
>
> **Решение: Фиксированный конвейер**
>
> Когда вам нужно, чтобы задачи выполнялись в **гарантированном, определенном порядке**, вы можете использовать `SequentialAgent`. Этот агент действует как сборочный конвейер, запуская каждого субагента в точном порядке, в котором вы их перечислили. Вывод одного агента автоматически становится вводом для следующего, создавая предсказуемый и надежный рабочий процесс.
>
> **Используйте Sequential (последовательность), когда:** Порядок имеет значение, вам нужен линейный конвейер или каждый шаг опирается на предыдущий.
>
> Чтобы узнать больше, ознакомьтесь с документацией по [последовательным агентам в ADK](https://google.github.io/adk-docs/agents/workflow-agents/sequential-agents/).
>
> **Архитектура: Конвейер создания поста в блог**

#### Пример: Создание поста в блог

````
# Агент-структуризатор: Создает первоначальный план поста в блоге.
outline_agent = Agent(
    name="OutlineAgent",
    model="gemini-2.5-flash-lite",
    instruction="""Создайте план поста в блоге по заданной теме, включающий:
    1. Цепляющий заголовок
    2. Вводный хук (крючок)
    3. 3-5 основных разделов с 2-3 пунктами для каждого
    4. Заключительную мысль""",
    output_key="blog_outline", # Результат работы этого агента будет сохранен в состоянии сессии с этим ключом.
)

print("✅ outline_agent создан.")
````

````
# Агент-писатель: Пишет полный пост в блоге на основе плана от предыдущего агента.
writer_agent = Agent(
    name="WriterAgent",
    model="gemini-2.5-flash-lite",
    # Плейсхолдер `{blog_outline}` автоматически внедряет значение состояния из вывода предыдущего агента.
    instruction="""Строго следуя этому плану: {blog_outline}
    Напишите краткий пост для блога объемом от 200 до 300 слов в увлекательном и информативном тоне.""",
    output_key="blog_draft", # Результат работы этого агента будет сохранен с этим ключом.
)

print("✅ writer_agent создан.")
````

````
# Агент-редактор: Редактирует и улучшает черновик от агента-писателя.
editor_agent = Agent(
    name="EditorAgent",
    model="gemini-2.5-flash-lite",
    # Этот агент получает `{blog_draft}` из вывода агента-писателя.
    instruction="""Отредактируйте этот черновик: {blog_draft}
    Ваша задача — улучшить текст, исправив грамматические ошибки, улучшив плавность и структуру предложений, а также повысив общую ясность.""",
    output_key="final_blog", # Это окончательный вывод всего конвейера.
)

print("✅ editor_agent создан.")
````

> Затем мы объединяем агентов под управлением последовательного агента (SequentialAgent), который запускает агентов в том порядке, в котором они перечислены:

````
root_agent = SequentialAgent(
    name="BlogPipeline",
    sub_agents=[outline_agent, writer_agent, editor_agent],
)

print("✅ Sequential Agent создан.")
````

> Давайте запустим агента и дадим ему тему для написания поста в блоге:

````
runner = InMemoryRunner(agent=root_agent)
response = await runner.run_debug("Напиши пост в блог о преимуществах многоагентных систем для разработчиков ПО")
````

> 👏 Отличная работа\! Теперь вы создали надежный «сборочный конвейер», используя последовательного агента, где каждый шаг выполняется в предсказуемом порядке.
>
> **Это идеально подходит для задач, которые строятся друг на друге, но это медленно, если задачи независимы.** Далее мы рассмотрим, как запускать несколько агентов одновременно, чтобы ускорить рабочий процесс.

**Раздел 4: Параллельные рабочие процессы**

>
>
> -----
>
> ### Раздел 4
>
> ## 🛣️ Параллельные рабочие процессы (Parallel) — Независимые исследователи
>
> **Проблема: Узкое место**
>
> Предыдущий последовательный агент хорош, но это сборочная линия. Каждый шаг должен ждать завершения предыдущего. Что, если у вас есть несколько задач, которые **не зависят** друг от друга? Например, исследование трех *разных* тем. Запуск их последовательно был бы медленным и неэффективным, создавая узкое место, где каждая задача ждет без необходимости.
>
> **Решение: Конкурентное выполнение**
>
> Когда у вас есть независимые задачи, вы можете запустить их все одновременно, используя `ParallelAgent`. Этот агент выполняет всех своих субагентов конкурентно (параллельно), значительно ускоряя рабочий процесс. Как только все параллельные задачи будут завершены, вы можете передать их объединенные результаты на финальный этап «агрегации».
>
> **Используйте Parallel (параллельность), когда:** Задачи независимы, скорость имеет значение, и вы можете выполнять их конкурентно.
>
> Чтобы узнать больше, ознакомьтесь с документацией по [параллельным агентам в ADK](https://google.github.io/adk-docs/agents/workflow-agents/parallel-agents/).
>
> **Архитектура: Исследование нескольких тем**

````
# Медицинский исследователь: Сосредоточен на медицинских прорывах.
health_researcher = Agent(
    name="HealthResearcher",
    model="gemini-2.5-flash-lite",
    instruction="""Исследуйте недавние медицинские прорывы. Включите 3 значительных достижения,
их практическое применение и предполагаемые сроки. Сделайте отчет кратким (100 слов).""",
    tools=[google_search],
    output_key="health_research", # Результат будет сохранен с этим ключом.
)

print("✅ health_researcher создан.")
````

````
# Финансовый исследователь: Сосредоточен на трендах финтеха.
finance_researcher = Agent(
    name="FinanceResearcher",
    model="gemini-2.5-flash-lite",
    instruction="""Исследуйте текущие тренды в финтехе. Включите 3 ключевых тренда,
их влияние на рынок и прогноз на будущее. Сделайте отчет кратким (100 слов).""",
    tools=[google_search],
    output_key="finance_research", # Результат будет сохранен с этим ключом.
)

print("✅ finance_researcher создан.")
````

````
# AggregatorAgent запускается *после* параллельного этапа для синтеза результатов.
aggregator_agent = Agent(
    name="AggregatorAgent",
    model="gemini-2.5-flash-lite",
    # Он использует плейсхолдеры для внедрения выводов параллельных агентов, которые теперь находятся в состоянии сессии.
    instruction="""Объедините эти три результата исследований в единое резюме для руководства:

    **Технологические тренды:**
    {tech_research}

    **Медицинские прорывы:**
    {health_research}

    **Инновации в финансах:**
    {finance_research}

    Ваше резюме должно выделять общие темы, удивительные связи и самые важные ключевые выводы из всех трех отчетов. Итоговое резюме должно быть около 200 слов.""",
    output_key="executive_summary", # Это будет финальный вывод всей системы.
)

print("✅ aggregator_agent создан.")
````

> 👉 **Затем мы объединяем агентов под управлением параллельного агента, который сам вложен внутри последовательного агента.**
>
> Этот дизайн гарантирует, что агенты-исследователи сначала запускаются параллельно, а затем, как только все их исследования завершены, агент-агрегатор собирает все результаты исследований в единый отчет:

````
# ParallelAgent запускает всех своих субагентов одновременно.
parallel_research_team = ParallelAgent(
    name="ParallelResearchTeam",
    sub_agents=[tech_researcher, health_researcher, finance_researcher],
)

# Этот SequentialAgent определяет высокоуровневый рабочий процесс: сначала запустить параллельную команду, затем запустить агрегатор.
root_agent = SequentialAgent(
    name="ResearchSystem",
    sub_agents=[parallel_research_team, aggregator_agent],
)

print("✅ Parallel и Sequential агенты созданы.")
````

> Давайте запустим агента и дадим ему промпт для исследования заданных тем:

````
runner = InMemoryRunner(agent=root_agent)
response = await runner.run_debug("Запусти ежедневный брифинг для руководства по технологиям, здоровью и финансам")
````

> 🎉 Отлично\! Вы увидели, как параллельные агенты могут значительно ускорить рабочие процессы за счет одновременного выполнения независимых задач.
>
> До сих пор все наши рабочие процессы запускались от начала до конца и затем останавливались. **Но что, если вам нужно просмотреть и улучшить результат несколько раз?** Далее мы создадим рабочий процесс, который может зацикливаться и улучшать собственную работу.

**Раздел 5: Циклические рабочие процессы**

>
>
> -----
>
> ### Раздел 5
>
> ## ➰ Циклические рабочие процессы (Loop) — Цикл доработки
>
> **Проблема: Качество с первой попытки**
>
> Все рабочие процессы, которые мы видели до сих пор, выполняются от начала до конца. `SequentialAgent` и `ParallelAgent` выдают свой окончательный результат и затем останавливаются. Этот подход «с одной попытки» не подходит для задач, требующих доработки и контроля качества. Что, если первый черновик нашей истории плох? У нас нет способа просмотреть его и попросить переписать.
>
> **Решение: Итеративная доработка**
>
> Когда задачу необходимо улучшать через циклы обратной связи и проверки, вы можете использовать `LoopAgent`. `LoopAgent` запускает набор субагентов повторно *до тех пор, пока не будет выполнено определенное условие или не будет достигнуто максимальное количество итераций.* Это создает цикл доработки, позволяя системе агентов улучшать свою работу снова и снова.
>
> **Используйте Loop (цикл), когда:** Требуется итеративное улучшение, важна доработка качества или нужны повторяющиеся циклы.
>
> Чтобы узнать больше, ознакомьтесь с документацией по [циклическим агентам в ADK](https://google.github.io/adk-docs/agents/workflow-agents/loop-agents/).
>
> **Архитектура: Написание истории и цикл критики**

#### Пример: Итеративная доработка истории

````
# Этот агент запускается ОДИН РАЗ в начале для создания первого черновика.
initial_writer_agent = Agent(
    name="InitialWriterAgent",
    model="gemini-2.5-flash-lite",
    instruction="""Основываясь на промпте пользователя, напишите первый черновик рассказа (около 100-150 слов).
    Выведите только текст рассказа, без вступления или объяснений.""",
    output_key="current_story", # Сохраняет первый черновик в состоянии.
)

print("✅ initial_writer_agent создан.")
````

````
# Единственная задача этого агента — предоставить обратную связь или сигнал одобрения. У него нет инструментов.
critic_agent = Agent(
    name="CriticAgent",
    model="gemini-2.5-flash-lite",
    instruction="""Вы — конструктивный критик историй. Просмотрите историю, предоставленную ниже.
    История: {current_story}

    Оцените сюжет истории, персонажей и темп.
    - Если история написана хорошо и завершена, вы ДОЛЖНЫ ответить точной фразой: "APPROVED"
    - В противном случае предоставьте 2-3 конкретных, действенных предложения по улучшению.""",
    output_key="critique", # Сохраняет отзыв в состоянии.
)

print("✅ critic_agent создан.")
````

> Теперь нам нужен способ, чтобы цикл действительно останавливался на основе отзыва критика. Сам по себе `LoopAgent` не знает автоматически, что "APPROVED" означает «стоп».
>
> Нам нужен агент, который даст явный сигнал завершить цикл.
>
> Мы делаем это в две части:
>
> 1.  Простая функция Python, которую `LoopAgent` понимает как сигнал «выхода».
> 2.  Агент, который может вызвать эту функцию, когда выполнено правильное условие.
>
> Сначала вы определите функцию `exit_loop`:

````
# Это функция, которую вызовет RefinerAgent для выхода из цикла.
def exit_loop():
    """Вызывайте эту функцию ТОЛЬКО когда критика равна 'APPROVED', указывая, что история закончена и больше изменений не требуется."""
    return {"status": "approved", "message": "История одобрена. Выход из цикла доработки."}

print("✅ exit_loop функция создана.")
````

> Чтобы позволить агенту вызвать эту функцию Python, мы оборачиваем её в `FunctionTool`. Затем мы создаем `RefinerAgent`, у которого есть этот инструмент.
>
> 👉 **Обратите внимание на его инструкции:** этот агент является «мозгом» цикла. Он читает `{critique}` от `CriticAgent` и решает, (1) вызвать ли инструмент `exit_loop` или (2) переписать историю.

````
# Этот агент дорабатывает историю на основе критики ИЛИ вызывает функцию выхода из цикла.
refiner_agent = Agent(
    name="RefinerAgent",
    model="gemini-2.5-flash-lite",
    instruction="""Вы — доработчик историй. У вас есть черновик истории и критика.

    Черновик истории: {current_story}
    Критика: {critique}

    Ваша задача — проанализировать критику.
    - ЕСЛИ критика В ТОЧНОСТИ равна "APPROVED", вы ДОЛЖНЫ вызвать функцию `exit_loop` и ничего больше.
    - В ПРОТИВНОМ СЛУЧАЕ, перепишите черновик истории, чтобы полностью учесть замечания из критики.""",

    output_key="current_story", # Он перезаписывает историю новой, улучшенной версией.
    tools=[FunctionTool(exit_loop)], # Инструмент теперь корректно инициализирован ссылкой на функцию.
)

print("✅ refiner_agent создан.")
````

> Затем мы объединяем агентов под управлением циклического агента, который сам вложен внутри последовательного агента.
>
> Этот дизайн гарантирует, что система сначала создает начальный черновик истории, затем запускается цикл доработки до указанного количества `max_iterations`:

````
# LoopAgent содержит агентов, которые будут запускаться повторно: Критик -> Доработчик.
story_refinement_loop = LoopAgent(
    name="StoryRefinementLoop",
    sub_agents=[critic_agent, refiner_agent],
    max_iterations=2, # Предотвращает бесконечные циклы
)

# Корневой агент — это SequentialAgent, который определяет общий рабочий процесс: Начальное написание -> Цикл доработки.
root_agent = SequentialAgent(
    name="StoryPipeline",
    sub_agents=[initial_writer_agent, story_refinement_loop],
)

print("✅ Loop и Sequential агенты созданы.")
````

> Давайте запустим агента и дадим ему тему для написания рассказа:

````
runner = InMemoryRunner(agent=root_agent)
response = await runner.run_debug("Напиши рассказ о смотрителе маяка, который обнаруживает таинственную светящуюся карту")
````

> Вы реализовали цилического агента, создав сложную систему, которая может итеративно проверять и улучшать собственный вывод. Это ключевой шаблон для обеспечения высококачественных результатов.
>
> Теперь у вас есть полный набор инструментов шаблонов рабочих процессов. Давайте соберем всё вместе и рассмотрим, как выбрать правильный шаблон для вашего случая использования.
